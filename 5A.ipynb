```python
# ============================================================
# Finding Minimum and Maximum using Divide and Conquer
# ============================================================
#
# Compare:
# 1. Naive approach
# 2. Divide and Conquer approach
#
# Array sizes:
# 10, 100, 1000, 10000
#
# The program counts the number of comparisons performed
# by both approaches.
# ============================================================

import random
import csv


# ------------------------------------------------------------
# Generate Random Array
# ------------------------------------------------------------

def generate_array(size):
    random.seed(42)
    return [random.randint(1, 100000) for _ in range(size)]


# ------------------------------------------------------------
# 1. Naive Approach
# ------------------------------------------------------------

def naive_min_max(arr):
    comparisons = 0

    minimum = arr[0]
    maximum = arr[0]

    for i in range(1, len(arr)):

        # Compare for minimum
        comparisons += 1
        if arr[i] < minimum:
            minimum = arr[i]

        # Compare for maximum
        comparisons += 1
        if arr[i] > maximum:
            maximum = arr[i]

    return minimum, maximum, comparisons


# ------------------------------------------------------------
# 2. Divide and Conquer Approach
# ------------------------------------------------------------

def divide_conquer_min_max(arr, low, high):

    # Base case: only one element
    if low == high:
        return arr[low], arr[low], 0

    # Base case: two elements
    if high == low + 1:

        comparisons = 1

        if arr[low] < arr[high]:
            return arr[low], arr[high], comparisons
        else:
            return arr[high], arr[low], comparisons

    # Find middle
    mid = (low + high) // 2

    # Solve left half
    left_min, left_max, left_comparisons = divide_conquer_min_max(
        arr, low, mid
    )

    # Solve right half
    right_min, right_max, right_comparisons = divide_conquer_min_max(
        arr, mid + 1, high
    )

    # Compare minimum values
    comparisons = 1
    if left_min < right_min:
        minimum = left_min
    else:
        minimum = right_min

    # Compare maximum values
    comparisons += 1
    if left_max > right_max:
        maximum = left_max
    else:
        maximum = right_max

    total_comparisons = (
        left_comparisons
        + right_comparisons
        + comparisons
    )

    return minimum, maximum, total_comparisons


# ------------------------------------------------------------
# Run Experiment
# ------------------------------------------------------------

def run_experiment():

    array_sizes = [10, 100, 1000, 10000]

    results = []

    print("=" * 80)
    print("MIN-MAX USING DIVIDE AND CONQUER")
    print("=" * 80)

    print(
        f"\n{'Array Size':<15}"
        f"{'Naive Comparisons':<22}"
        f"{'D&C Comparisons':<22}"
        f"{'Naive Min/Max':<20}"
        f"{'D&C Min/Max'}"
    )

    print("-" * 100)

    for size in array_sizes:

        arr = generate_array(size)

        # Naive method
        naive_min, naive_max, naive_comparisons = naive_min_max(arr)

        # Divide and Conquer method
        dc_min, dc_max, dc_comparisons = divide_conquer_min_max(
            arr, 0, len(arr) - 1
        )

        # Verify that both methods give the same result
        if naive_min != dc_min or naive_max != dc_max:
            print("ERROR: Results do not match!")

        print(
            f"{size:<15}"
            f"{naive_comparisons:<22}"
            f"{dc_comparisons:<22}"
            f"{str(naive_min) + '/' + str(naive_max):<20}"
            f"{str(dc_min) + '/' + str(dc_max)}"
        )

        results.append({
            "Array Size": size,
            "Naive Comparisons": naive_comparisons,
            "Divide and Conquer Comparisons": dc_comparisons,
            "Minimum": dc_min,
            "Maximum": dc_max
        })

    return results


# ------------------------------------------------------------
# Save Results to CSV
# ------------------------------------------------------------

def save_results(results):

    filename = "min_max_results.csv"

    with open(filename, "w", newline="") as file:

        fieldnames = [
            "Array Size",
            "Naive Comparisons",
            "Divide and Conquer Comparisons",
            "Minimum",
            "Maximum"
        ]

        writer = csv.DictWriter(
            file,
            fieldnames=fieldnames
        )

        writer.writeheader()
        writer.writerows(results)

    print(f"\nResults saved to: {filename}")


# ------------------------------------------------------------
# Complexity Analysis
# ------------------------------------------------------------

def display_complexity():

    print("\n" + "=" * 80)
    print("COMPLEXITY ANALYSIS")
    print("=" * 80)

    print("""
Naive Approach:
    Time Complexity  : O(n)
    Space Complexity : O(1)

Divide and Conquer:
    Time Complexity  : O(n)
    Space Complexity : O(log n) due to recursion

Comparison Count:

Naive:
    2(n - 1)

Divide and Conquer:
    Approximately 3n/2 - 2 comparisons
    for suitable array sizes.

Therefore, both approaches have O(n) time complexity,
but Divide and Conquer performs fewer comparisons.
""")


# ------------------------------------------------------------
# Main Function
# ------------------------------------------------------------

def main():

    results = run_experiment()

    save_results(results)

    display_complexity()

    print("\n" + "=" * 80)
    print("EXPERIMENT COMPLETED SUCCESSFULLY")
    print("=" * 80)


# ------------------------------------------------------------
# Program Entry Point
# ------------------------------------------------------------

if __name__ == "__main__":
    main()
```
