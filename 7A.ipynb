```python
# ============================================================
# N-Queens Problem using Backtracking
# ============================================================
#
# Tasks:
# 1. Find and display all solutions for N = 4
# 2. Count the total number of solutions for N = 8
#
# ============================================================


# ------------------------------------------------------------
# Check if a Queen can be placed
# ------------------------------------------------------------

def is_safe(board, row, col):

    # Check previous rows
    for previous_row in range(row):

        previous_col = board[previous_row]

        # Same column
        if previous_col == col:
            return False

        # Same diagonal
        if abs(previous_row - row) == abs(previous_col - col):
            return False

    return True


# ------------------------------------------------------------
# Backtracking Function
# ------------------------------------------------------------

def solve_n_queens(n):

    board = [-1] * n

    solutions = []

    def backtrack(row):

        # All queens have been placed
        if row == n:
            solutions.append(board.copy())
            return

        # Try every column in the current row
        for col in range(n):

            if is_safe(board, row, col):

                # Place queen
                board[row] = col

                # Move to next row
                backtrack(row + 1)

                # Remove queen
                board[row] = -1

    backtrack(0)

    return solutions


# ------------------------------------------------------------
# Display Chessboard
# ------------------------------------------------------------

def display_board(solution):

    n = len(solution)

    for row in range(n):

        for col in range(n):

            if solution[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()

    print()


# ------------------------------------------------------------
# Display All N=4 Solutions
# ------------------------------------------------------------

def display_n4_solutions():

    solutions = solve_n_queens(4)

    print("\n" + "=" * 50)
    print("N = 4 : ALL SOLUTIONS")
    print("=" * 50)

    print(f"\nTotal solutions for N = 4: {len(solutions)}")

    for index, solution in enumerate(solutions, start=1):

        print(f"\nSolution {index}:")
        display_board(solution)


# ------------------------------------------------------------
# Count N=8 Solutions
# ------------------------------------------------------------

def count_n8_solutions():

    solutions = solve_n_queens(8)

    print("\n" + "=" * 50)
    print("N = 8 : TOTAL NUMBER OF SOLUTIONS")
    print("=" * 50)

    print(f"\nTotal solutions for N = 8: {len(solutions)}")


# ------------------------------------------------------------
# Main Function
# ------------------------------------------------------------

def main():

    print("=" * 60)
    print("N-QUEENS PROBLEM USING BACKTRACKING")
    print("=" * 60)

    # Display all solutions for N = 4
    display_n4_solutions()

    # Count solutions for N = 8
    count_n8_solutions()

    # Complexity
    print("\n" + "=" * 60)
    print("COMPLEXITY ANALYSIS")
    print("=" * 60)

    print("""
Theoretical worst-case time complexity:
    O(N!)

Space complexity:
    O(N)

The algorithm places one queen per row and uses
backtracking whenever the current placement is invalid.
""")


# ------------------------------------------------------------
# Program Entry Point
# ------------------------------------------------------------

if __name__ == "__main__":
    main()
```
