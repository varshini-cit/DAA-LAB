```python
# ============================================================
# Single Source Shortest Path using Dijkstra's Algorithm
# ============================================================

import heapq


# ------------------------------------------------------------
# Graph Configuration
# ------------------------------------------------------------

VERTICES = 6

# Directed weighted graph
# (source, destination, weight)

EDGES = [
    (0, 1, 4),
    (0, 2, 2),
    (1, 2, 1),
    (1, 3, 5),
    (2, 1, 1),
    (2, 3, 8),
    (2, 4, 10),
    (3, 4, 2),
    (3, 5, 6),
    (4, 5, 3)
]

SOURCE = 0


# ------------------------------------------------------------
# Create Adjacency List
# ------------------------------------------------------------

def create_graph(vertices, edges):

    graph = [[] for _ in range(vertices)]

    for u, v, weight in edges:
        graph[u].append((v, weight))

    return graph


# ------------------------------------------------------------
# Dijkstra's Algorithm
# ------------------------------------------------------------

def dijkstra(graph, source):

    vertices = len(graph)

    # Initially, all distances are infinity
    distance = [float("inf")] * vertices

    # Parent array is used to reconstruct shortest paths
    parent = [-1] * vertices

    # Distance from source to itself is 0
    distance[source] = 0

    # Priority queue stores:
    # (distance, vertex)
    priority_queue = [(0, source)]

    while priority_queue:

        current_distance, current_vertex = heapq.heappop(
            priority_queue
        )

        # Ignore outdated entries
        if current_distance > distance[current_vertex]:
            continue

        # Visit all neighboring vertices
        for neighbor, weight in graph[current_vertex]:

            new_distance = current_distance + weight

            # Relaxation
            if new_distance < distance[neighbor]:

                distance[neighbor] = new_distance
                parent[neighbor] = current_vertex

                heapq.heappush(
                    priority_queue,
                    (new_distance, neighbor)
                )

    return distance, parent


# ------------------------------------------------------------
# Reconstruct Shortest Path
# ------------------------------------------------------------

def get_path(parent, destination):

    path = []

    current = destination

    while current != -1:
        path.append(current)
        current = parent[current]

    # Reverse the path to get source -> destination
    path.reverse()

    return path


# ------------------------------------------------------------
# Display Results
# ------------------------------------------------------------

def display_results(distance, parent, source):

    print("\n" + "=" * 65)
    print("DIJKSTRA'S SINGLE SOURCE SHORTEST PATH")
    print("=" * 65)

    print(f"\nSource Vertex: {source}")

    print("\n" + "-" * 65)
    print(f"{'Vertex':<12}{'Shortest Distance':<22}{'Shortest Path'}")
    print("-" * 65)

    for vertex in range(len(distance)):

        if distance[vertex] == float("inf"):
            print(f"{vertex:<12}{'INF':<22}No path")
            continue

        path = get_path(parent, vertex)

        path_string = " -> ".join(map(str, path))

        print(
            f"{vertex:<12}"
            f"{distance[vertex]:<22}"
            f"{path_string}"
        )


# ------------------------------------------------------------
# Main Function
# ------------------------------------------------------------

def main():

    print("=" * 65)
    print("SINGLE SOURCE SHORTEST PATH USING DIJKSTRA'S ALGORITHM")
    print("=" * 65)

    print("\nNumber of vertices:", VERTICES)
    print("Source vertex:", SOURCE)

    print("\nDirected Weighted Edges:")

    for u, v, weight in EDGES:
        print(f"{u} -> {v}  Weight = {weight}")

    # Create graph
    graph = create_graph(VERTICES, EDGES)

    # Run Dijkstra's algorithm
    distance, parent = dijkstra(graph, SOURCE)

    # Display shortest paths
    display_results(distance, parent, SOURCE)

    # --------------------------------------------------------
    # Complexity Analysis
    # --------------------------------------------------------

    print("\n" + "=" * 65)
    print("TIME COMPLEXITY")
    print("=" * 65)

    print("""
Using an adjacency list and a priority queue:

Time Complexity : O((V + E) log V)

Where:
    V = number of vertices
    E = number of edges

Space Complexity: O(V + E)
""")


# ------------------------------------------------------------
# Program Entry Point
# ------------------------------------------------------------

if __name__ == "__main__":
    main()
```
