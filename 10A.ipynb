```python
# ============================================================
# Improving Quick Sort Efficiency using Randomized Algorithm
# ============================================================
#
# Algorithms:
# 1. Deterministic Quick Sort
#    - Pivot = last element
#
# 2. Randomized Quick Sort
#    - Pivot = randomly selected element
#
# Array size = 10000
#
# Configurations:
# 1. Random array
# 2. Sorted array
# 3. Reverse sorted array
# 4. Array with duplicate values
#
# Measures:
# - Execution time
# - Number of comparisons
# ============================================================

import random
import time
import sys


# Increase recursion limit for worst-case Quick Sort
sys.setrecursionlimit(100000)


# ------------------------------------------------------------
# Deterministic Quick Sort
# Pivot = Last Element
# ------------------------------------------------------------

def deterministic_partition(arr, low, high, counter):

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        counter[0] += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def deterministic_quick_sort(arr, low, high, counter):

    if low < high:

        pivot_index = deterministic_partition(
            arr,
            low,
            high,
            counter
        )

        deterministic_quick_sort(
            arr,
            low,
            pivot_index - 1,
            counter
        )

        deterministic_quick_sort(
            arr,
            pivot_index + 1,
            high,
            counter
        )


# ------------------------------------------------------------
# Randomized Quick Sort
# Pivot = Random Element
# ------------------------------------------------------------

def randomized_partition(arr, low, high, counter):

    # Select random pivot
    random_index = random.randint(low, high)

    # Move random pivot to the last position
    arr[random_index], arr[high] = (
        arr[high],
        arr[random_index]
    )

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        counter[0] += 1

        if arr[j] <= pivot:

            i += 1

            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = (
        arr[high],
        arr[i + 1]
    )

    return i + 1


def randomized_quick_sort(arr, low, high, counter):

    if low < high:

        pivot_index = randomized_partition(
            arr,
            low,
            high,
            counter
        )

        randomized_quick_sort(
            arr,
            low,
            pivot_index - 1,
            counter
        )

        randomized_quick_sort(
            arr,
            pivot_index + 1,
            high,
            counter
        )


# ------------------------------------------------------------
# Generate Test Arrays
# ------------------------------------------------------------

def generate_test_arrays(n):

    # 1. Random array
    random_array = [
        random.randint(1, 100000)
        for _ in range(n)
    ]

    # 2. Already sorted array
    sorted_array = list(range(n))

    # 3. Reverse sorted array
    reverse_sorted_array = list(range(n, 0, -1))

    # 4. Array containing many duplicate values
    duplicate_array = [
        random.randint(1, 10)
        for _ in range(n)
    ]

    return {
        "Random Array": random_array,
        "Sorted Array": sorted_array,
        "Reverse Sorted Array": reverse_sorted_array,
        "Duplicate Values": duplicate_array
    }


# ------------------------------------------------------------
# Measure Deterministic Quick Sort
# ------------------------------------------------------------

def measure_deterministic(arr):

    test_array = arr.copy()

    counter = [0]

    start_time = time.perf_counter()

    deterministic_quick_sort(
        test_array,
        0,
        len(test_array) - 1,
        counter
    )

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return execution_time, counter[0], test_array


# ------------------------------------------------------------
# Measure Randomized Quick Sort
# ------------------------------------------------------------

def measure_randomized(arr):

    test_array = arr.copy()

    counter = [0]

    start_time = time.perf_counter()

    randomized_quick_sort(
        test_array,
        0,
        len(test_array) - 1,
        counter
    )

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return execution_time, counter[0], test_array


# ------------------------------------------------------------
# Main Experiment
# ------------------------------------------------------------

def main():

    N = 10000

    random.seed(42)

    test_arrays = generate_test_arrays(N)

    print("=" * 95)
    print("QUICK SORT: DETERMINISTIC VS RANDOMIZED")
    print("=" * 95)

    print(f"\nArray Size: {N}")

    print("\nComparison Count:")
    print("Each comparison is one comparison between an array")
    print("element and the selected pivot.")

    print("\n" + "-" * 95)

    print(
        f"{'Configuration':<22}"
        f"{'Algorithm':<22}"
        f"{'Time (seconds)':<20}"
        f"{'Comparisons':<20}"
    )

    print("-" * 95)

    results = []

    for configuration, original_array in test_arrays.items():

        # ----------------------------------------------------
        # Deterministic Quick Sort
        # ----------------------------------------------------

        (
            deterministic_time,
            deterministic_comparisons,
            deterministic_sorted
        ) = measure_deterministic(original_array)

        # ----------------------------------------------------
        # Randomized Quick Sort
        # ----------------------------------------------------

        (
            randomized_time,
            randomized_comparisons,
            randomized_sorted
        ) = measure_randomized(original_array)

        # Verify both algorithms sorted correctly
        expected = sorted(original_array)

        if deterministic_sorted != expected:
            print("ERROR: Deterministic Quick Sort failed!")

        if randomized_sorted != expected:
            print("ERROR: Randomized Quick Sort failed!")

        print(
            f"{configuration:<22}"
            f"{'Deterministic':<22}"
            f"{deterministic_time:<20.6f}"
            f"{deterministic_comparisons:<20}"
        )

        print(
            f"{'':<22}"
            f"{'Randomized':<22}"
            f"{randomized_time:<20.6f}"
            f"{randomized_comparisons:<20}"
        )

        print("-" * 95)

        results.append(
            (
                configuration,
                deterministic_time,
                deterministic_comparisons,
                randomized_time,
                randomized_comparisons
            )
        )

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    print("\n" + "=" * 95)
    print("SUMMARY")
    print("=" * 95)

    for result in results:

        (
            configuration,
            deterministic_time,
            deterministic_comparisons,
            randomized_time,
            randomized_comparisons
        ) = result

        print(f"\n{configuration}")

        print(
            f"  Deterministic : "
            f"{deterministic_time:.6f} sec, "
            f"{deterministic_comparisons} comparisons"
        )

        print(
            f"  Randomized    : "
            f"{randomized_time:.6f} sec, "
            f"{randomized_comparisons} comparisons"
        )

    # --------------------------------------------------------
    # Complexity Analysis
    # --------------------------------------------------------

    print("\n" + "=" * 95)
    print("COMPLEXITY ANALYSIS")
    print("=" * 95)

    print("""
Deterministic Quick Sort:
    Best Case    : O(n log n)
    Average Case : O(n log n)
    Worst Case   : O(n²)

Randomized Quick Sort:
    Expected     : O(n log n)
    Worst Case   : O(n²)

Space Complexity:
    Average      : O(log n)
    Worst Case   : O(n)

The randomized algorithm reduces the probability of
repeatedly selecting a poor pivot, especially for inputs
such as already sorted or reverse-sorted arrays.
""")


# ------------------------------------------------------------
# Program Entry Point
# ------------------------------------------------------------

if __name__ == "__main__":
    main()
```
