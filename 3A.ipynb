```python
# ============================================================
# Implementation of Kruskal's and Prim's Algorithms
# for Minimum Spanning Tree (MST)
# ============================================================

import heapq


# ------------------------------------------------------------
# Graph Configuration
# ------------------------------------------------------------

VERTICES = 7

# Each edge is represented as:
# (source, destination, weight)

EDGES = [
    (0, 1, 4),
    (0, 2, 3),
    (1, 2, 1),
    (1, 3, 2),
    (2, 3, 4),
    (3, 4, 2),
    (4, 5, 6),
    (4, 6, 3),
    (5, 6, 5)
]


# ------------------------------------------------------------
# Disjoint Set Union (DSU) for Kruskal's Algorithm
# ------------------------------------------------------------

class DisjointSet:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, vertex):
        if self.parent[vertex] != vertex:
            self.parent[vertex] = self.find(self.parent[vertex])

        return self.parent[vertex]

    def union(self, u, v):

        root_u = self.find(u)
        root_v = self.find(v)

        if root_u == root_v:
            return False

        if self.rank[root_u] < self.rank[root_v]:
            self.parent[root_u] = root_v

        elif self.rank[root_u] > self.rank[root_v]:
            self.parent[root_v] = root_u

        else:
            self.parent[root_v] = root_u
            self.rank[root_u] += 1

        return True


# ------------------------------------------------------------
# Kruskal's Algorithm
# ------------------------------------------------------------

def kruskal(vertices, edges):

    # Sort edges according to their weights
    sorted_edges = sorted(edges, key=lambda edge: edge[2])

    dsu = DisjointSet(vertices)

    mst = []
    total_weight = 0

    for u, v, weight in sorted_edges:

        # Add edge only if it does not create a cycle
        if dsu.union(u, v):

            mst.append((u, v, weight))
            total_weight += weight

            # MST contains V-1 edges
            if len(mst) == vertices - 1:
                break

    return mst, total_weight


# ------------------------------------------------------------
# Prim's Algorithm
# ------------------------------------------------------------

def prim(vertices, edges):

    # Create adjacency list
    graph = [[] for _ in range(vertices)]

    for u, v, weight in edges:
        graph[u].append((weight, v))
        graph[v].append((weight, u))

    visited = [False] * vertices

    # Priority queue:
    # (weight, current_vertex, parent_vertex)
    priority_queue = [(0, 0, -1)]

    mst = []
    total_weight = 0

    while priority_queue and len(mst) < vertices - 1:

        weight, current, parent = heapq.heappop(priority_queue)

        if visited[current]:
            continue

        visited[current] = True

        # Ignore the starting vertex
        if parent != -1:
            mst.append((parent, current, weight))
            total_weight += weight

        # Add adjacent edges to priority queue
        for edge_weight, neighbor in graph[current]:

            if not visited[neighbor]:
                heapq.heappush(
                    priority_queue,
                    (edge_weight, neighbor, current)
                )

    return mst, total_weight


# ------------------------------------------------------------
# Display MST
# ------------------------------------------------------------

def display_mst(name, mst, total_weight):

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print("\nEdges in MST:")

    for u, v, weight in mst:
        print(f"{u} -- {v}  Weight = {weight}")

    print(f"\nTotal Minimum Weight = {total_weight}")


# ------------------------------------------------------------
# Main Function
# ------------------------------------------------------------

def main():

    print("=" * 60)
    print("MINIMUM SPANNING TREE - KRUSKAL vs PRIM")
    print("=" * 60)

    print("\nNumber of vertices:", VERTICES)
    print("Number of edges:", len(EDGES))

    print("\nGraph Edges:")

    for u, v, weight in EDGES:
        print(f"{u} -- {v}  Weight = {weight}")

    # Run Kruskal
    kruskal_mst, kruskal_weight = kruskal(
        VERTICES,
        EDGES
    )

    # Run Prim
    prim_mst, prim_weight = prim(
        VERTICES,
        EDGES
    )

    # Display results
    display_mst(
        "KRUSKAL'S ALGORITHM",
        kruskal_mst,
        kruskal_weight
    )

    display_mst(
        "PRIM'S ALGORITHM",
        prim_mst,
        prim_weight
    )

    # --------------------------------------------------------
    # Verification
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("VERIFICATION")
    print("=" * 60)

    print(f"\nKruskal MST Weight : {kruskal_weight}")
    print(f"Prim MST Weight    : {prim_weight}")

    if kruskal_weight == prim_weight:
        print("\n✓ Both algorithms produce the same minimum weight.")
    else:
        print("\n✗ The MST weights are different.")

    # --------------------------------------------------------
    # Complexity Analysis
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print("TIME COMPLEXITY")
    print("=" * 60)

    print("""
Kruskal's Algorithm:
    Sorting edges      : O(E log E)
    DSU operations     : O(E α(V))
    Overall            : O(E log E)

Prim's Algorithm:
    Priority Queue     : O(E log V)
    Overall            : O(E log V)

Where:
    V = number of vertices
    E = number of edges
    α(V) = inverse Ackermann function
""")


# ------------------------------------------------------------
# Program Entry Point
# ------------------------------------------------------------

if __name__ == "__main__":
    main()
```
