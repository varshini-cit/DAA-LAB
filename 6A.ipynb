```python
# ============================================================
# Matrix Chain Multiplication using Dynamic Programming
# ============================================================
#
# Matrices:
# A1 = 10 x 30
# A2 = 30 x 5
# A3 = 5 x 60
# A4 = 60 x 10
#
# Find:
# 1. Optimal multiplication order
# 2. Minimum number of scalar multiplications
# ============================================================


# ------------------------------------------------------------
# Matrix dimensions
# ------------------------------------------------------------
# If matrices are:
#
# A1 = 10 x 30
# A2 = 30 x 5
# A3 = 5 x 60
# A4 = 60 x 10
#
# dimensions = [10, 30, 5, 60, 10]

dimensions = [10, 30, 5, 60, 10]

number_of_matrices = len(dimensions) - 1


# ------------------------------------------------------------
# Matrix Chain Multiplication
# ------------------------------------------------------------

def matrix_chain_order(dimensions):

    n = len(dimensions) - 1

    # dp[i][j] stores minimum multiplication cost
    # for matrices Ai through Aj
    dp = [
        [0 for _ in range(n + 1)]
        for _ in range(n + 1)
    ]

    # split[i][j] stores the position where the optimal
    # multiplication is split
    split = [
        [0 for _ in range(n + 1)]
        for _ in range(n + 1)
    ]

    # chain_length is the number of matrices in the chain
    for chain_length in range(2, n + 1):

        for i in range(1, n - chain_length + 2):

            j = i + chain_length - 1

            # Start with infinity
            dp[i][j] = float("inf")

            # Try every possible split position
            for k in range(i, j):

                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + dimensions[i - 1]
                    * dimensions[k]
                    * dimensions[j]
                )

                # Keep the minimum cost
                if cost < dp[i][j]:

                    dp[i][j] = cost
                    split[i][j] = k

    return dp, split


# ------------------------------------------------------------
# Construct Optimal Parenthesization
# ------------------------------------------------------------

def construct_optimal_order(split, i, j):

    # Only one matrix
    if i == j:
        return f"A{i}"

    k = split[i][j]

    left = construct_optimal_order(
        split,
        i,
        k
    )

    right = construct_optimal_order(
        split,
        k + 1,
        j
    )

    return f"({left} x {right})"


# ------------------------------------------------------------
# Display DP Table
# ------------------------------------------------------------

def display_dp_table(dp, n):

    print("\n" + "=" * 70)
    print("DP TABLE - MINIMUM MULTIPLICATION COST")
    print("=" * 70)

    print("\nRows and columns represent matrix indices.")

    print("\n       ", end="")

    for j in range(1, n + 1):
        print(f"{'A' + str(j):>10}", end="")

    print()

    print("-" * (10 * (n + 1)))

    for i in range(1, n + 1):

        print(f"{'A' + str(i):<7}", end="")

        for j in range(1, n + 1):

            if i > j:
                print(f"{'-':>10}", end="")
            else:
                print(f"{dp[i][j]:>10}", end="")

        print()


# ------------------------------------------------------------
# Display Split Table
# ------------------------------------------------------------

def display_split_table(split, n):

    print("\n" + "=" * 70)
    print("SPLIT TABLE")
    print("=" * 70)

    print("\nEach value represents the optimal split position k.")

    print("\n       ", end="")

    for j in range(1, n + 1):
        print(f"{'A' + str(j):>10}", end="")

    print()

    print("-" * (10 * (n + 1)))

    for i in range(1, n + 1):

        print(f"{'A' + str(i):<7}", end="")

        for j in range(1, n + 1):

            if i >= j:
                print(f"{'-':>10}", end="")
            else:
                print(f"{split[i][j]:>10}", end="")

        print()


# ------------------------------------------------------------
# Main Function
# ------------------------------------------------------------

def main():

    print("=" * 70)
    print("MATRIX CHAIN MULTIPLICATION USING DYNAMIC PROGRAMMING")
    print("=" * 70)

    print("\nMatrices:")

    for i in range(number_of_matrices):

        rows = dimensions[i]
        columns = dimensions[i + 1]

        print(
            f"A{i + 1} = {rows} x {columns}"
        )

    # Calculate optimal order
    dp, split = matrix_chain_order(dimensions)

    # Construct parenthesization
    optimal_order = construct_optimal_order(
        split,
        1,
        number_of_matrices
    )

    minimum_cost = dp[1][number_of_matrices]

    # Display results
    print("\n" + "=" * 70)
    print("RESULT")
    print("=" * 70)

    print(f"\nOptimal multiplication order:")
    print(optimal_order)

    print(
        f"\nMinimum number of scalar multiplications: "
        f"{minimum_cost}"
    )

    # Display tables
    display_dp_table(
        dp,
        number_of_matrices
    )

    display_split_table(
        split,
        number_of_matrices
    )

    # Complexity
    print("\n" + "=" * 70)
    print("COMPLEXITY ANALYSIS")
    print("=" * 70)

    print("""
Time Complexity  : O(n^3)
Space Complexity : O(n^2)

Where n is the number of matrices.
""")


# ------------------------------------------------------------
# Program Entry Point
# ------------------------------------------------------------

if __name__ == "__main__":
    main()
```
