```python
# ============================================================
# Efficient Bin Packing using Approximation Algorithms
# ============================================================
#
# Algorithms:
# 1. First Fit (FF)
# 2. First Fit Decreasing (FFD)
# 3. Best Fit (BF)
#
# Item weights:
# [0.5, 0.7, 0.3, 0.9, 0.2,
#  0.6, 0.8, 0.4, 0.1, 0.5]
#
# Bin capacity = 1.0
#
# ============================================================

import math


# ------------------------------------------------------------
# Input
# ------------------------------------------------------------

ITEMS = [
    0.5, 0.7, 0.3, 0.9, 0.2,
    0.6, 0.8, 0.4, 0.1, 0.5
]

BIN_CAPACITY = 1.0


# ------------------------------------------------------------
# First Fit
# ------------------------------------------------------------

def first_fit(items, capacity):

    bins = []

    for item in items:

        placed = False

        # Try existing bins from left to right
        for i in range(len(bins)):

            if bins[i] + item <= capacity + 1e-9:

                bins[i] += item
                placed = True
                break

        # If item cannot fit into any existing bin,
        # create a new bin
        if not placed:
            bins.append(item)

    return bins


# ------------------------------------------------------------
# First Fit Decreasing
# ------------------------------------------------------------

def first_fit_decreasing(items, capacity):

    # Sort items in decreasing order
    sorted_items = sorted(items, reverse=True)

    return first_fit(sorted_items, capacity)


# ------------------------------------------------------------
# Best Fit
# ------------------------------------------------------------

def best_fit(items, capacity):

    bins = []

    for item in items:

        best_bin = -1
        minimum_remaining_space = float("inf")

        for i in range(len(bins)):

            remaining_space = bins[i] + item

            if remaining_space <= capacity + 1e-9:

                # Select the bin that will have the
                # smallest remaining space
                remaining_after_placement = (
                    capacity - remaining_space
                )

                if remaining_after_placement < minimum_remaining_space:

                    minimum_remaining_space = (
                        remaining_after_placement
                    )

                    best_bin = i

        if best_bin != -1:

            bins[best_bin] += item

        else:

            bins.append(item)

    return bins


# ------------------------------------------------------------
# Calculate Lower Bound
# ------------------------------------------------------------

def calculate_lower_bound(items, capacity):

    total_weight = sum(items)

    return math.ceil(
        total_weight / capacity - 1e-9
    )


# ------------------------------------------------------------
# Display Bin Arrangement
# ------------------------------------------------------------

def display_bins(bins):

    for i, total_weight in enumerate(bins, start=1):

        print(
            f"Bin {i}: "
            f"Total weight = {total_weight:.2f}, "
            f"Remaining capacity = "
            f"{BIN_CAPACITY - total_weight:.2f}"
        )


# ------------------------------------------------------------
# Run Experiment
# ------------------------------------------------------------

def main():

    print("=" * 75)
    print("EFFICIENT BIN PACKING USING APPROXIMATION ALGORITHMS")
    print("=" * 75)

    print("\nItems:")
    print(ITEMS)

    print(f"\nBin Capacity: {BIN_CAPACITY:.1f}")

    total_weight = sum(ITEMS)

    print(f"Total Item Weight: {total_weight:.1f}")

    # Calculate theoretical lower bound
    lower_bound = calculate_lower_bound(
        ITEMS,
        BIN_CAPACITY
    )

    print(
        f"Theoretical Lower Bound: "
        f"{lower_bound} bins"
    )

    # --------------------------------------------------------
    # First Fit
    # --------------------------------------------------------

    ff_bins = first_fit(
        ITEMS,
        BIN_CAPACITY
    )

    print("\n" + "=" * 75)
    print("1. FIRST FIT (FF)")
    print("=" * 75)

    display_bins(ff_bins)

    print(f"\nNumber of bins used: {len(ff_bins)}")

    # --------------------------------------------------------
    # First Fit Decreasing
    # --------------------------------------------------------

    ffd_bins = first_fit_decreasing(
        ITEMS,
        BIN_CAPACITY
    )

    print("\n" + "=" * 75)
    print("2. FIRST FIT DECREASING (FFD)")
    print("=" * 75)

    print(
        "\nItems sorted in decreasing order:"
    )

    print(sorted(ITEMS, reverse=True))

    display_bins(ffd_bins)

    print(f"\nNumber of bins used: {len(ffd_bins)}")

    # --------------------------------------------------------
    # Best Fit
    # --------------------------------------------------------

    bf_bins = best_fit(
        ITEMS,
        BIN_CAPACITY
    )

    print("\n" + "=" * 75)
    print("3. BEST FIT (BF)")
    print("=" * 75)

    display_bins(bf_bins)

    print(f"\nNumber of bins used: {len(bf_bins)}")

    # --------------------------------------------------------
    # Comparison
    # --------------------------------------------------------

    print("\n" + "=" * 75)
    print("COMPARISON")
    print("=" * 75)

    print(
        f"\n{'Algorithm':<25}"
        f"{'Bins Used':<15}"
        f"{'Lower Bound':<15}"
        f"{'Extra Bins'}"
    )

    print("-" * 70)

    algorithms = [
        ("First Fit", len(ff_bins)),
        ("First Fit Decreasing", len(ffd_bins)),
        ("Best Fit", len(bf_bins))
    ]

    for name, bins_used in algorithms:

        extra_bins = bins_used - lower_bound

        print(
            f"{name:<25}"
            f"{bins_used:<15}"
            f"{lower_bound:<15}"
            f"{extra_bins}"
        )

    # --------------------------------------------------------
    # Final Observation
    # --------------------------------------------------------

    best_result = min(
        len(ff_bins),
        len(ffd_bins),
        len(bf_bins)
    )

    print("\n" + "=" * 75)
    print("OBSERVATION")
    print("=" * 75)

    print(
        f"\nThe theoretical lower bound is "
        f"{lower_bound} bins."
    )

    print(
        f"The best approximation result uses "
        f"{best_result} bins."
    )

    if best_result == lower_bound:

        print(
            "The best algorithm achieves the theoretical "
            "lower bound."
        )

    else:

        print(
            "The approximation algorithms use more bins "
            "than the theoretical lower bound."
        )

    # --------------------------------------------------------
    # Complexity
    # --------------------------------------------------------

    print("\n" + "=" * 75)
    print("TIME COMPLEXITY")
    print("=" * 75)

    print("""
First Fit:
    Time Complexity: O(n²) for this straightforward implementation

First Fit Decreasing:
    Sorting: O(n log n)
    Packing: O(n²)
    Overall: O(n²)

Best Fit:
    Time Complexity: O(n²) for this straightforward implementation
""")


# ------------------------------------------------------------
# Program Entry Point
# ------------------------------------------------------------

if __name__ == "__main__":
    main()
```
